# **Magnetic Anisotropy**

Magnetic anisotropy is the property of a magnetic material (especially ferromagnetic or ferrimagnetic ones) where its magnetic behavior depends on direction. In simple terms: it's easier to magnetize the material (or keep its magnetization) along certain directions than others.
In an isotropic magnetic material, magnetization behaves the same way no matter which direction you apply a magnetic field. But in anisotropic materials, there are preferred directions.

## **Magnetic Anisotropy and Spin–Orbit Coupling in SIESTA**

- Magnetism in density functional theory arises from the intrinsic spin of electrons, and a system becomes magnetic when the populations of spin-up and spin-down electrons differ, producing a net magnetic moment. In simple (collinear) calculations the spin is constrained to lie along a fixed axis, so only its magnitude and sign matter. However, this approximation cannot describe situations where the ***direction*** of magnetization influences the total energy. To capture such effects one must use a non-collinear formalism together with spin–orbit coupling, as implemented in SIESTA. Spin–orbit coupling links the electron’s spin angular momentum with its orbital motion around the nucleus, effectively coupling the magnetic moment to the crystal lattice. Because orbital motion depends on the local atomic environment, rotating the spin relative to the lattice slightly changes the electronic structure and therefore the total energy. This directional dependence of the total energy is called magnetic anisotropy, and the energy difference between preferred (easy-axis) and unfavorable (hard-axis) orientations is known as the magnetic anisotropy energy (MAE).

- In SIESTA, each atomic magnetic moment is treated as a three-dimensional vector whose magnitude and direction are specified using spherical coordinates through the `DM.InitSpin` block.

                                %block DM.InitSpin
                                atom index,  |S|, theta,  phi
                                1   4.0    0.0   0.0
                                2  -4.0   90.0  45.0
                                %endblock DM.InitSpin

- The magnitude gives the strength of the moment, while two angles define its orientation: the polar angle $\theta$ measures the tilt away from the z-axis, and the azimuthal angle $\phi$ specifies rotation within the xy-plane. These angles are internally converted into Cartesian components that enter the Hamiltonian and influence the self-consistent solution. To compute magnetic anisotropy, one performs several calculations with identical structural and numerical parameters but different spin orientations (for example along x, y, and z), then compares the resulting total energies. The lowest-energy direction identifies the easy axis, and the energy differences quantify the MAE, revealing how strongly the lattice constrains the magnetization direction.

# Analysis of Magnetic Anisotropy in Pt₂ using SIESTA

## 1. Compare the total energies for the different alignments of spin moment and the dimer axis. In which cases are the total energies the same? What do these cases have in common?

To analyze the magnetic anisotropy, we computed the total energy of the Pt₂ dimer for several configurations obtained by either rotating the spin direction while keeping the molecular geometry fixed, or by rotating the molecular axis while keeping the spin direction fixed. When comparing these energies, we observe that some configurations give identical total energies within small numerical fluctuations (of the order of less than 1 meV). These cases correspond to situations where the ***relative orientation*** between the spin moment and the dimer axis is the same.

This indicates that the total energy does not depend on the absolute Cartesian direction (x, y, or z), but only on the angle between the spin moment and the molecular axis. For example, configurations where the spin is parallel to the bond direction produce the same energy, independent of whether the bond lies along x, y, or z. Similarly, all perpendicular configurations also give approximately the same energy. Therefore, the common feature of cases with equal energies is that they share the same spin–bond angle (either parallel or perpendicular). This confirms that the relevant physical quantity controlling the energy is the spin orientation relative to the lattice, not the global coordinate system.

---

## 2. How large is the magnetic anisotropy energy (the change in total energy depending on the spin direction)?

The magnetic anisotropy energy (MAE) is defined as the difference between the highest and lowest total energies obtained for different spin orientations:

$[
MAE = E_{\text{hard}} - E_{\text{easy}}
]$

Using the calculated total energies for different spin directions (while keeping the geometry fixed), we compute the energy differences and convert them into meV. The resulting values are of the order of a few meV. This small energy scale is expected for magnetic anisotropy, since it originates from spin–orbit coupling, which is typically much weaker than the total bonding energy of the system (which is on the order of eV).

Hence, the MAE for the Pt₂ dimer is in the meV range, indicating that only a small energy cost is required to rotate the magnetization direction. The orientation with the lowest total energy corresponds to the easy axis of magnetization, while higher-energy orientations correspond to hard axes.

---

## 3. What would happen if Spin non-colinear was used instead of Spin spin-orbit? Would the same values be obtained?

If we perform the calculations using `Spin non-colinear` instead of `Spin spin-orbit`, the spin–orbit coupling term is removed from the Hamiltonian. In this case, the system becomes rotationally invariant with respect to spin orientation, meaning that the total energy no longer depends on the direction of the spin moment.

Physically, without spin–orbit coupling there is no interaction that couples the spin direction to the lattice. Therefore, rotating the spin does not change the electronic structure or the total energy. As a result, all spin orientations would produce exactly the same total energy (up to numerical noise), and the magnetic anisotropy energy would be zero.

Thus, magnetic anisotropy arises entirely from spin–orbit coupling, and without it the anisotropy disappears.
