# Optimization the second zeta matching radii

From the previous exercise , we found that lowest enrgy when radial cut off at Hs, O2s, 02p are equal with 4.5,5.5 and 7.5 respectively at individual exprement . So in these exercise we're going to fixed this radial cut off value and find the lowest energy by varying the radial messoff value

The above one is the PAO.Basis BLock that we used in these tutorial. For this tutorial we can used slightly different fdf file as given in the manual since we substitute the PAO.Basis block which there already in  with the above new one. This will us more helpfull and clean workspace environment.

### Changing the value of H1s radial matching radii

```python
#check directory where we are
!pwd
```

```python
!cp water.fdf water.template.fdf
```

```python
!vi water.template.fdf
```

```python
match = [1.5, 2.0, 2.5, 3.0]

for M in match:
    folder = f"H1s_match_{M}"

    !mkdir -p {folder}

    # replace matching radius only
    !sed "s/MATCH/{M}/" water.template.fdf > {folder}/water.fdf

    # copy pseudopotentials
    !cp H.psml O.psml {folder}/

    # run siesta
    !cd {folder} && siesta < water.fdf > water.out
```

```python
for M in match:
    print(f"H1s_match_{M}", end="  ")
    !grep -F "E + p_basis" H1s_match_{M}/water.out
```
