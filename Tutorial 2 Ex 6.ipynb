# Optimization of Polarization Orbitals

### Finding the matching radii of polarization orbitals of Hydrogen atom that give lowest binding energy

```python
!pwd
```

```python
!ls
```

```python
!cp water.fdf water.template_H.fdf
```

```python
!vi water.template_H.fdf
```

```python
pmatch = [3.5, 4.0, 4.5, 5.0, 5.5, 6.0]

for M in pmatch:
    folder = f"H_2p_{M}"

    !mkdir -p {folder}
    !sed "s/pMATCH/{M}/" water.template_H.fdf > {folder}/water.fdf
    !cp H.psml O.psml {folder}/
    !cd {folder} && siesta < water.fdf > water.out
```

```python
!pwd
```

```python
pmatch = [3.5, 4.0, 4.5, 5.0, 5.5, 6.0]
for M in pmatch:
    print(f"H_2p_{M}", end="  ")
    !grep -F "E + p_basis" H_2p_{M}/water.out 
    !grep "SCF cycle converged" H_2p_{M}/water.out
    
```
