# Exercise-1.3 Change the SplitNorm and check the Enthalpy, Total Energy and Orbital volume

```python
%cd /home/l-rishiraj/siesta-docs/work-files/tutorials/basic/basis-optimization/01-Eshift_SplitNorm/
```

```python
!mkdir ex_3
```

```python
!cp H.psml  O.psml  water.fdf ex_3/
```

```python
!ls ex_3
```

```python
%cd ex_3
```

```python
splitnorm = [0.45, 0.55, 0.65]
EnergyShift = 0.001  # Ry (fixed)

for SN in splitnorm:
    folder = f"SN_{SN}"

    # create folder
    !mkdir -p {folder}

    # copy input files
    !cp water.fdf O.psml H.psml {folder}/

    # modify SplitNorm (dimensionless!)
    !sed -i "s/PAO.SplitNorm.*/PAO.SplitNorm {SN}/" {folder}/water.fdf

    # fix EnergyShift
    !sed -i "s/PAO.EnergyShift.*/PAO.EnergyShift {EnergyShift} Ry/" {folder}/water.fdf

    # run siesta
    !cd {folder} && siesta water.fdf > water.out
```

```python
for SN in splitnorm:
    print(f"SN _{SN}", end="  ")
    !grep -F "E + p_basis" SN_{SN}*/water.out
```

```python
for SN in splitnorm:
    print(f"SN_{SN}",end =" ")
    !grep -F "Total =" SN_{SN}*/water.out
```

```python
for SN in splitnorm:
    print(f"SN_{SN}", end=" ")
    !cat SN_{SN}/*BASIS_ENTHALPY
```
