```python
%%bash

# --- Your desired sequence (same as loop order) ---
KLIST=("3x3x2" "5x5x3" "6x6x4" "8x8x6" "10x10x8" "12x12x10")

# --- Header (nice aligned columns) ---
printf "%-15s %-10s %-15s %-18s %-15s\n" "#Folder" "Nkpts" "SCF_iters" "TotalEnergy(eV)" "WallTime(s)" > graphite_kconv.dat
# --- Extract data in correct order ---
for k in "${KLIST[@]}"; do

    dir="KGRID_${k}"
    out="$dir/graphite.out"

    if [ ! -f "$out" ]; then
        echo "Missing $out"
        continue
    fi

    # ---- Number of k-points ----
    Nk=$(grep -i "Number of k-points" "$out" | tail -n 1 | awk '{print $NF}')

    # ---- SCF iterations (better pattern) ----
    SCF=$(grep -c "scf" "$out")

    # ---- Total energy ----
    E=$(grep -i " Total = " "$out" | tail -n 1 | awk '{print $4}')

    # ---- Wall time (try multiple patterns) ----
    WT=$(grep -i "Wall" "$out" | tail -n 1 | awk '{print $NF}')
    if [ -z "$WT" ]; then
        WT=$(grep -i "Elapsed time" "$out" | tail -n 1 | awk '{print $NF}')
    fi
    if [ -z "$WT" ]; then
        WT="NA"
    fi

    # ---- Write aligned row ----
    printf "%-15s %-10s %-17s %-18s %-15s\n" "$dir" "$Nk" "$SCF" "$E" "$WT" >> graphite_kconv.dat
    
done

echo " Saved: graphite_kconv.dat"
echo ""
cat graphite_kconv.dat
```

```python
%%bash

KLIST=("3x3x2" "5x5x3" "6x6x4" "8x8x6" "10x10x8" "12x12x10")

# Reference energy from last k-grid
last_k="${KLIST[-1]}"
last_out="KGRID_${last_k}/graphite.out"
Eref=$(grep -i " Total = " "$last_out" | tail -n 1 | awk '{print $4}')

printf "%-14s %-8s %-10s %-18s %-14s %-14s %-18s\n" \
"#Folder" "Nkpts" "SCF_iters" "TotalEnergy(eV)" "WallTime(s)" "CPUTime(s)" "EnergyError(eV)" \
> graphite_kconv.dat

for k in "${KLIST[@]}"; do

    dir="KGRID_${k}"
    out="$dir/graphite.out"

    if [ ! -f "$out" ]; then
        echo "Missing $out"
        continue
    fi

    # ---- Number of k-points ----
    Nk=$(grep -i "Number of k-points" "$out" | tail -n 1 | awk '{print $NF}')

    # ---- SCF iterations ----
    SCF=$(grep -i -c "scf: Iter" "$out")
    if [ "$SCF" -eq 0 ]; then
        SCF=$(grep -i -c "scf:" "$out")
    fi

    # ---- Total energy ----
    E=$(grep -i " Total = " "$out" | tail -n 1 | awk '{print $4}')

    # ---- Wall time ----
    WT=$(grep -i "Elapsed wall time (sec)" "$out" | tail -n 1 | awk -F"=" '{print $2}' | xargs)
    

    # ---- CPU time (sum Tot.time column in CPU table) ----
    CPU=$(awk '
    /timer: CPU execution times \(sec\):/ {flag=1; next}
    flag==1 && $1=="Routine" {next}
    flag==1 && NF>=4 {
        if ($4 ~ /^[0-9.]+$/) sum += $4
    }
    flag==1 && $0 ~ /^timer:/ {flag=0}
    END {
        if (sum>0) printf "%.6f", sum;
        else print "NA"
    }' "$out")

    # ---- Energy error ----
    dE=$(awk -v e="$E" -v ref="$Eref" 'BEGIN{printf "%.10f", (e-ref)}')

    printf "%-14s %-8s %-10s %-18s %-14s %-14s %-18s\n" \
    "$dir" "$Nk" "$SCF" "$E" "$WT" "$CPU" "$dE" >> graphite_kconv.dat

done

echo " Saved: graphite_kconv.dat"
echo ""
cat graphite_kconv.dat
```

## Finding from the tutorial

1) Graphite needs dense k-points in x–y plane

Graphite has strong sp² bonding in the basal plane, so electrons disperse strongly along kx and ky.
That’s why increasing 3×3 → 12×12 changes the total energy a lot.

2) Coarse k-grid gives wrong total energy

With very few k-points (like 3×3×2), Brillouin zone integration is poor.
So many electronic states are missed, giving a large energy error (~0.9 eV).

3) Convergence occurs around 8×8×2

After 8×8×2, the energy difference between higher grids becomes very small (≈ meV scale).
This means the Brillouin-zone sampling is sufficient and energy is nearly stable.

4) EnergyError sign changes are normal

Energy may overshoot or undershoot the converged value depending on k-point sampling.
This happens because k-point integration is a numerical approximation, not strictly monotonic.

5) SCF iterations indicate difficulty of electronic convergence

Graphite is a semi-metal, so SCF convergence can be harder due to Fermi-level sensitivity.
That’s why some grids (like 6×6×2) show high SCF iterations (43).

6) CPU/Wall time increases with Nkpts

More k-points means solving Kohn–Sham equations for more points in the Brillouin zone.
So computational cost scales roughly with Nkpts × SCF_iters, increasing wall and CPU time.

A k-point convergence test was performed for graphite by systematically increasing the Monkhorst–Pack grid and monitoring the total energy, wall time, and CPU time. The EnergyError decreases rapidly when moving from coarse sampling (3×3×2) to denser k-point meshes, indicating that insufficient Brillouin-zone sampling produces large inaccuracies in the calculated total energy. Beyond approximately the 8×8×2 grid, the EnergyError becomes very small and nearly constant, showing that the total energy is well converged with respect to k-points. In contrast, both CPUTime and WallTime generally increase with the number of k-points, since more k-points require additional electronic structure calculations in each SCF cycle. Therefore, an 8×8×2 k-grid provides a good balance between computational efficiency and accuracy, while denser meshes such as 12×12×2 may be used for higher precision calculations.
