# Create a total energy file and Bond length

```python
%cd /home/l-rishiraj/siesta-docs/work-files/tutorials/basic/first-encounter/02-CH4-Basis/exercise_2
```

```python
%%bash

echo "# EnergyShift(meV)   TotalEnergy(eV)" > energy_SZ.dat

for d in SZ_*; do
  shift=${d#SZ_}
  energy=$(grep "Total " $d/ch4.out | tail -1 | awk '{print $NF}')
  echo "$shift   $energy" >> energy_SZ.dat
done
```

```python
!cat energy_SZ.dat
```

```python
%%bash
echo "# EnergyShift(meV)  Bondlength(Ang)" > bond_SZ.dat
for d in SZ_*; do
  shift=${d#SZ_}
  bond=$(awk '
    $1=="C" {cx=$2; cy=$3; cz=$4}
    $2=="H" {
      dx=$2-cx; dy=$3-cy; dz=$4-cz;
      print sqrt(dx*dx + dy*dy + dz*dz);
      exit
    }
  ' "$d/ch4.STRUCT_OUT")
  echo "$shift  $bond" >> bond_SZ.dat
done
```

```python
!cat bond_SZ.dat
```

```python
energy_shift = [10, 50, 100, 150, 200, 250, 300]
basis_set = ["SZ", "DZ", "SZP", "TZP", "DZP"]

for BS in basis_set:
    for ES in energy_shift:
        folder = f"{BS}_{ES}"

        # create folder
        !mkdir -p {folder}

        # copy input files
        !cp ch4.fdf C.psml H.psml {folder}/

        # modify basis size
        !sed -i "s/PAO.BasisSize.*/PAO.BasisSize {BS}/" {folder}/ch4.fdf

        # modify EnergyShift
        !sed -i "s/PAO.EnergyShift.*/PAO.EnergyShift {ES} meV/" {folder}/ch4.fdf

        # run siesta
        !cd {folder} && siesta ch4.fdf > ch4.out
```

```python
for BS in basis_set:
    print(f"\n{BS}")
    for ES in energy_shift:
        print(f"{ES} meV :", end=" ")

        # show total energy
        !grep "Total =" {BS}_{ES}/ch4.out | tail -1

        # show SCF convergence message
        !grep "SCF.*converged" {BS}_{ES}/ch4.out
```
